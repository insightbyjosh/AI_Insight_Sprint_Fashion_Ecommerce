# 5. Data Preparation & Feature Engineering

## Purpose

After cleaning the datasets, new analytical variables are created to enhance business reporting and support executive decision-making.

These engineered features simplify analysis, improve reporting efficiency, and provide the foundation for revenue, customer, product, and marketing intelligence throughout the notebook.

### Create Time-Based Features
### Business Reason

Executives rarely analyse individual transaction dates. Instead, they ask questions like:

Which month generated the highest revenue?
Which quarter performed best?
Are there seasonal trends?

We'll create those fields now.

```python
# Create Year
sales["year"] = sales["sale_date"].dt.year

# Create Month Number
sales["month_number"] = sales["sale_date"].dt.month

# Create Month Name
sales["month_name"] = sales["sale_date"].dt.month_name()

# Create Year-Month
sales["month"] = sales["sale_date"].dt.to_period("M").astype(str)

# Create Quarter
sales["quarter"] = sales["sale_date"].dt.quarter

sales.head()
```

### Review

These variables will be reused throughout the notebook. Instead of repeatedly extracting dates in every analysis section, we've created them once and can reference them consistently.

### Create Discount Labels

```python
sales["discount_status"] = sales["discounted"].map({
    0: "No Discount",
    1: "Discount Applied"
})

sales.head()
```

### Create Product Profit Metrics

### Business Reason

Revenue alone doesn't tell us whether products are profitable. By calculating profit metrics now, we'll be able to evaluate products from both a sales and profitability perspective later.

```python
products["gross_profit"] = (
    products["catalog_price"]
    - products["cost_price"]
)

products["gross_margin_%"] = (
    products["gross_profit"]
    / products["catalog_price"]
    * 100
).round(2)

products.head()
```

### Create Campaign Duration

```python
campaigns["campaign_duration_days"] = (
    campaigns["end_date"]
    - campaigns["start_date"]
).dt.days + 1

campaigns.head()
```

## Revenue by Sales Channel

### Objective

This analysis evaluates the performance of each sales channel by comparing total revenue, average order value, transaction volume, and revenue contribution.

The objective is to identify the most effective sales channel and provide recommendations for improving channel performance and resource allocation.

```python
# ==========================================
# Revenue by Sales Channel
# ==========================================

channel_revenue = (
    sales.groupby("channel")
    .agg(
        Total_Revenue=("total_amount", "sum"),
        Average_Order_Value=("total_amount", "mean"),
        Transactions=("sale_id", "count")
    )
    .round(2)
)

# Revenue Share
channel_revenue["Revenue_Share_%"] = (
    channel_revenue["Total_Revenue"]
    / channel_revenue["Total_Revenue"].sum()
    * 100
).round(2)

# Revenue per Transaction
channel_revenue["Revenue_per_Transaction"] = (
    channel_revenue["Total_Revenue"]
    / channel_revenue["Transactions"]
).round(2)

channel_revenue = channel_revenue.sort_values(
    by="Total_Revenue",
    ascending=False
)

channel_revenue
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(
    channel_revenue.index,
    channel_revenue["Total_Revenue"]
)

plt.title("Revenue by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Total Revenue")

plt.tight_layout()
plt.show()
```

### Business Insight

The E-commerce channel generated **171,675.72** in revenue, representing **52.95%** of total sales, while App Mobile contributed **152,560.94**, accounting for **47.05%**.

Although both channels performed well, E-commerce generated **19,114.78** more revenue than App Mobile and also achieved a slightly higher Average Order Value (**362.95** versus **353.15**).

The relatively balanced revenue distribution indicates that the business has successfully diversified its sales channels, reducing dependence on a single platform while maintaining strong online performance.

### Management Recommendation

Management should continue investing in the E-commerce platform because it currently delivers the highest revenue and average order value.

At the same time, App Mobile should be strengthened through user experience improvements, personalized recommendations, and targeted promotional campaigns to increase transaction values and narrow the performance gap.

### Expected Business Impact

## Monthly Revenue Trend

### Objective

This analysis evaluates monthly revenue performance to identify sales trends, growth patterns, and periods of strong or weak business performance.

The objective is to determine whether revenue is improving over time and to identify opportunities for corrective action where necessary.

```python
# ==========================================
# Monthly Revenue Performance
# ==========================================

monthly_revenue = (
    sales.groupby("month")
    .agg(
        Total_Revenue=("total_amount", "sum"),
        Average_Order_Value=("total_amount", "mean"),
        Transactions=("sale_id", "count")
    )
    .round(2)
)

monthly_revenue["Revenue_Growth_%"] = (
    monthly_revenue["Total_Revenue"]
    .pct_change() * 100
).round(2)

monthly_revenue
```

## 15.4 Campaign Duration Analysis

### Objective

This analysis evaluates the duration of marketing campaigns to understand the company's campaign planning strategy and promotional scheduling.

The objective is to identify the average campaign length, compare campaign durations, and support future marketing planning decisions.

```python
# ==========================================
# Campaign Duration Analysis
# ==========================================

# Convert dates to datetime
campaigns["start_date"] = pd.to_datetime(campaigns["start_date"])
campaigns["end_date"] = pd.to_datetime(campaigns["end_date"])

# Calculate campaign duration
campaigns["Campaign_Duration_Days"] = (
    campaigns["end_date"] - campaigns["start_date"]
).dt.days

campaign_duration = campaigns[[
    "campaign_name",
    "channel",
    "Campaign_Duration_Days"
]]

campaign_duration
```

```python
duration_summary = pd.DataFrame({

    "Metric":[
        "Average Campaign Duration",
        "Shortest Campaign",
        "Longest Campaign"
    ],

    "Value":[
        campaigns["Campaign_Duration_Days"].mean().round(2),
        campaigns["Campaign_Duration_Days"].min(),
        campaigns["Campaign_Duration_Days"].max()
    ]

})

duration_summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
    campaigns["campaign_name"],
    campaigns["Campaign_Duration_Days"]
)

plt.xticks(rotation=45, ha="right")

plt.title("Campaign Duration by Marketing Campaign")
plt.xlabel("Campaign")
plt.ylabel("Duration (Days)")

plt.tight_layout()
plt.show()
```

The company's seven marketing campaigns ranged from **6 to 11 days**, with an **average campaign duration of 8 days**.

Most campaigns were conducted over approximately one week, indicating a structured and consistent promotional schedule rather than highly variable campaign lengths.

The shortest campaign, **Spring Flash Sale**, lasted **6 days**, while **TIVA Week** was the longest campaign at **11 days**. This suggests that the company adjusts campaign duration based on the nature and objectives of individual promotions while maintaining an overall balanced campaign strategy.

### Management Recommendation

Management should continue maintaining a structured campaign schedule while evaluating campaign performance against duration. Future campaigns should compare customer engagement, conversion rates, and revenue generated to determine the optimal campaign length for different promotional objectives.

Where campaign performance data becomes available, management should test shorter and longer campaign durations to identify the most effective balance between customer engagement and marketing efficiency.

Optimising campaign duration based on measurable business outcomes is expected to improve campaign effectiveness, enhance customer engagement, increase marketing efficiency, and maximise the return on promotional investments.
